## 2.1 Dasar Teori

Perulangan for menjalankan blok kode untuk setiap nilai dalam suatu urutan. Fungsi `range()` sering digunakan untuk menghasilkan urutan bilangan, sedangkan nested for digunakan ketika setiap iterasi luar perlu menjalankan iterasi dalam.

### 2.1.1 Perulangan for

Perulangan for menjalankan blok kode satu kali untuk setiap nilai yang diberikan. Variabel iterasi menerima nilai berikutnya pada setiap putaran.

```python
for i in range(1, 4):
    print(i)
```

### 2.1.2 Fungsi range()

`range()` menghasilkan urutan bilangan untuk perulangan. Bentuk `range(stop)` dimulai dari 0, sedangkan `range(start, stop)` dimulai dari `start` dan berhenti sebelum mencapai `stop`.

```python
for angka in range(1, 4):
    print(angka)
```

### 2.1.3 Nested for

Nested for adalah perulangan for yang berada di dalam perulangan lain. Untuk setiap satu iterasi perulangan luar, perulangan dalam dijalankan sampai selesai. Struktur ini umum digunakan untuk membentuk keluaran berupa baris dan kolom.

```python
for baris in range(2):
    for kolom in range(3):
        print("*", end="")
    print()
```

### 2.1.4 Membentuk Pola dengan Perulangan Bersarang

Perulangan luar biasanya mengatur baris, sedangkan perulangan dalam mengatur isi setiap baris. Banyaknya iterasi perulangan dalam dapat dibuat bergantung pada nilai iterasi luar untuk menghasilkan pola segitiga atau bentuk lainnya.

### 2.1.5 Masukan dan Batas Iterasi

Nilai dari pengguna dapat menentukan banyaknya iterasi atau ukuran pola. Karena batas akhir pada `range()` tidak ikut diproses, batas perlu disesuaikan, misalnya dengan menambahkan 1 ketika angka batas ingin disertakan.

## 2.2 Kode Program

Bagian ini membahas program latihan Pekan 5. Contoh mencakup perulangan for, penggunaan `range()`, nested for untuk membentuk pola, penjumlahan berulang, serta pencetakan piramida.

### 2.2.1 Program "perulangan_for1_2611532019"

Program meminta jumlah perulangan, lalu menampilkan nomor iterasi dari 1 sampai nilai masukan.

````python
ulang_2019 = int(input("Masukkan jumlah perulangan: "))

for i in range(1, ulang_2019 + 1):
    print(f"Perulangan ke-{i}")
````

### 2.2.2 Program "perulangan_for2_2611532019"

Program membandingkan dua bentuk penggunaan `range()`: iterasi yang dimulai dari 0 dan iterasi yang dimulai dari 1 hingga batas masukan.

````python
ulang_2019 = int(input("Masukkan jumlah perulangan: "))

print("Perulangan dengan range(ulang_2019):")
for i in range(ulang_2019):
    print(i, end="")
print()

print("Perulangan dari 1 sampai batas masukan:")
for i in range(1, ulang_2019 + 1):
    print(i, end="")
print()
````

`range(ulang_2019)` menghasilkan nilai mulai dari 0 dan berhenti sebelum `ulang_2019`. Bentuk `range(1, ulang_2019 + 1)` menghasilkan hitungan yang dimulai dari 1 dan menyertakan batas masukan.

### 2.2.3 Program "nested_for1_2611532019"

Perulangan luar mengatur nomor baris, sedangkan perulangan dalam mencetak titik sebelum nilai baris. Jumlah titik berkurang pada setiap baris.

Kode berikut menggunakan dua perulangan bersarang untuk membentuk pola sesuai batas yang dimasukkan.

````python
batas_2019 = int(input("Masukkan nilai batas: "))

for line_2019 in range(1, batas_2019 + 1):
    for j in range(1, (-line_2019 + batas_2019) + 1):
        print(".", end="")
    print(line_2019)
````

Jumlah titik pada perulangan dalam ditentukan oleh nilai iterasi perulangan luar. Setelah titik dicetak, nilai `line_2019` ditampilkan sebagai penanda baris.

### 2.2.4 Program nested_for2_2611532019

Program memuat dua contoh nested for: membentuk persegi panjang dari karakter bintang dan membentuk tabel angka berdasarkan indeks baris serta kolom.

````python
batas_2019_1 = int(input("Masukkan nilai batas: "))
for line_2019 in range(1, batas_2019_1 + 1):
    for j in range(1, batas_2019_1 + 1):
        print("*", end="")
    print()

batas_2019_2 = int(input("Masukkan nilai batas: "))
for line_2019 in range(batas_2019_2 + 1):
    for j in range(batas_2019_2 + 1):
        print(line_2019 + j, end="")
    print()
````

Contoh pertama mengulang pencetakan bintang dengan jumlah kolom tetap. Contoh kedua mencetak hasil penjumlahan indeks baris dan kolom; setiap baris diproses oleh perulangan luar.

### 2.2.5 Program nested_for3_2611532019

Program membentuk pola bingkai dan simbol menggunakan nested for serta percabangan di dalam perulangan. Masukan tinggi harus berupa bilangan genap.

````python
tinggi_2019 = int(input("Masukkan tinggi pola: "))

if tinggi_2019 % 2 != 0:
    print("Masukkan bilangan genap")
else:
    lebar_2019 = (2 * tinggi_2019) - 2

    for i in range(1, tinggi_2019 + 1):
        c_2019 = i
        b_2019 = c_2019 + 1

        for j in range(1, lebar_2019 + 1):
            if i == 1 or i == tinggi_2019:
                if j == 1 or j == lebar_2019:
                    print("#", end="")
                else:
                    print("=", end="")
            elif j == 1 or j == lebar_2019:
                print("|", end="")
            elif j == c_2019:
                print("<", end="")
            elif j == b_2019:
                print(">", end="")
            elif j == lebar_2019 - c_2019:
                print("<", end="")
            elif c_2019 < j < lebar_2019 - c_2019:
                print("*", end="")
            elif j == lebar_2019 - c_2019 + 1:
                print(">", end="")
            else:
                print(" ", end="")
        print()
````

Perulangan luar mengatur baris pola dan perulangan dalam mengatur karakter pada setiap kolom. Percabangan menentukan karakter yang dicetak berdasarkan posisi baris dan kolom.

### 2.2.6 Program "perulangan_for3_2611532019"

Program menjumlahkan bilangan dari 1 hingga nilai masukan. Variabel `jumlah_2019` menyimpan hasil penjumlahan yang terus diperbarui pada setiap iterasi.

Kode Program:

````python
ulang_2019 = int(input("Masukkan jumlah perulangan: "))
jumlah_2019 = 0

for i in range(1, ulang_2019 + 1):
    print(i, end="")
    jumlah_2019 += i

    if i < ulang_2019:
        print("+", end="")
    else:
        print("=", jumlah_2019)

print()
print("Jumlah =", jumlah_2019)
````

### 2.2.7 Program tugas_prak5_2611532019

Program mencetak pola piramida bintang. Nested for pertama mencetak spasi di depan, sedangkan perulangan berikutnya mencetak kelompok bintang pada setiap baris.

````python
tinggi_2019 = int(input("Masukkan tinggi piramida: "))

for i in range(1, tinggi_2019 + 1):
    for j in range(1, tinggi_2019 - i + 1):
        print(" ", end="")
    for k in range(i):
        print("* ", end="")
    print()
````

Pola memiliki satu baris untuk setiap nilai `i`. Banyak spasi awal berkurang saat baris bertambah, sementara jumlah kelompok bintang meningkat.

## 2.3 Tugas Praktikum

Tugas praktikum Pekan 5 adalah program **"Jam Pasir Kristal Palindromik"**. Program menerima ukuran skala `N`, kemudian menggunakan perulangan untuk mencetak bingkai, bagian atas jam pasir, poros tengah, dan bagian bawah.

### 2.3.1 Kode Program "tugas5_2019"

Berikut kodenya :

````python
print("=== PROGRAM JAM PASIR KRISTAL PALINDROMIK (PEKAN 5) ===")
n_2019 = int(input("Masukkan ukuran skala jam pasir (N): "))

print("#", end="")
for i_2019 in range(4 * n_2019 + 5):
    print("=", end="")
print("#")

for baris_2019 in range(n_2019, 0, -1):
    print("| ", end="")
    for kiri_2019 in range(2 * (n_2019 - baris_2019)):
        print(" ", end="")
    for angka_2019 in range(baris_2019, 0, -1):
        print(angka_2019, end=" ")
    print("<*>", end="")
    for angka_2019 in range(1, baris_2019 + 1):
        print(" ", end="")
        print(angka_2019, end="")
    for kanan_2019 in range(2 * (n_2019 - baris_2019)):
        print(" ", end="")
    print(" |")

print("|", end="")
for kiri_2019 in range(2 * n_2019 + 1):
    print(" ", end="")
print("<*>", end="")
for kanan_2019 in range(2 * n_2019 + 1):
    print(" ", end="")
print("|")

for baris_2019 in range(1, n_2019 + 1):
    print("| ", end="")
    for kiri_2019 in range(2 * (n_2019 - baris_2019)):
        print(" ", end="")
    for angka_2019 in range(baris_2019, 0, -1):
        print(angka_2019, end=" ")
    print("<*>", end="")
    for angka_2019 in range(1, baris_2019 + 1):
        print(" ", end="")
        print(angka_2019, end="")
    for kanan_2019 in range(2 * (n_2019 - baris_2019)):
        print(" ", end="")
    print(" |")

print("#", end="")
for i_2019 in range(4 * n_2019 + 5):
    print("=", end="")
print("#")
````
