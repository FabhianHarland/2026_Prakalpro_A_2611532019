## 2.1 Dasar Teori
Praktikum ini berfokus pada pengenalan operator pada Python, pengenalan dasar sintaksis, serta logika pemrosesan awal dalam pemrograman Python. Untuk mendukung pelaksanaan praktikum, terdapat beberapa konsep dan landasan teori utama yang digunakan, antara lain:




### 2.1.1 Operator Aritmatika.
Operator Aritmatika digunakan untuk melakukan operasi matematika dasar seperti penjumlahan, pengurangan, perkalian, hingga perpangkatan.

| Operator | Deskripsi | Contoh |
| :---: | :--- | :---: |
| `+` | Penjumlahan | `a + b` |
| `-` | Pengurangan | `a - b` |
| `*` | Perkalian | `a * b` |
| `/` | Pembagian (hasil berupa *float*) | `a / b` |
| `%` | Sisa Bagi (*Modulus*) | `a % b` |
| `**` | Pangkat (*Exponentiation*) | `a ** b` |
| `//` | Pembagian Bulat (*Floor Division*) | `a // b` |

### 2.1.2 Operator Penugasan (Assignment).
Operator penugasan digunakan untuk memberikan atau memperbarui nilai ke dalam suatu variabel. Selain penugasan sederhana (=), Python mendukung operator penugasan gabungan untuk menyingkat operasi aritmatika.

| Operator | Contoh | Ekspresi Setara | Penjelasan |
| :---: | :---: | :---: | :--- |
| `=` | `x = 5` | `x = 5` | Mengisi nilai `5` ke dalam variabel `x` |
| `+=` | `x += 3` | `x = x + 3` | Menambahkan nilai `3` ke nilai `x` saat ini |
| `-=` | `x -= 2` | `x = x - 2` | Mengurangi nilai `x` saat ini dengan `2` |
| `*=` | `x *= 4` | `x = x * 4` | Mengalikan nilai `x` saat ini dengan `4` |
| `/=` | `x /= 2` | `x = x / 2` | Membagi nilai `x` saat ini dengan `2` (hasil *float*) |

### 2.1.3 Operator Bitwise.
Operator bitwise digunakan untuk melakukan operasi tingkat bit (biner) pada data bertipe bilangan bulat (integer). Operasi ini bekerja dengan membandingkan bit demi bit.

| Operator | Nama Operator | Deskripsi |
| :---: | :--- | :--- |
| `&` | **AND** | Bernilai `1` jika kedua bit bernilai `1` |
| `\|` | **OR** | Bernilai `1` jika salah satu bit bernilai `1` |
| `^` | **XOR** | Bernilai `1` jika bit yang dibandingkan berbeda |
| `~` | **NOT** | Membalik nilai bit (`0` menjadi `1`, dan sebaliknya) |
| `<<` | **Left Shift** | Menggeser bit ke kiri |
| `>>` | **Right Shift** | Menggeser bit ke kanan |

### 2.1.4 Operator Logika.
Operator logika digunakan untuk membandingkan dua ekspresi logis atau kondisi. Hasil dari pemrosesan operator ini adalah nilai kebenaran (Boolean), yaitu True atau False..

| Operator | Deskripsi | Contoh | Hasil |
| :---: | :--- | :---: | :---: |
| `and` | Bernilai `True` jika kedua kondisi bernilai `True` | `(5 > 2) and (3 < 8)` | `True` |
| `or` | Bernilai `True` jika salah satu kondisi bernilai `True` | `(5 > 2) or (3 > 8)` | `True` |
| `not` | Membalikkan nilai logika (`True` jadi `False`, dan sebaliknya) | `not(5 > 2)` | `False` |

### 2.1.5 Operator Keanggotaan dan Operator Identitas.
#### 2.1.5.1 Operator Keanggotaan.
Digunakan untuk menguji apakah suatu nilai atau elemen terdapat di dalam suatu struktur data koleksi (seperti list, tuple, string, atau dictionary).

| Operator | Deskripsi | Contoh Ekspresi | Hasil (`a = [1, 2, 3]`) |
| :---: | :--- | :---: | :---: |
| `in` | Bernilai `True` jika nilai ditemukan di dalam urutan/koleksi | `1 in a` | `True` |
| `not in` | Bernilai `True` jika nilai tidak ditemukan di dalam urutan/koleksi | `5 not in a` | `True` |

#### 2.1.5.2 Operator Identitas.
Digunakan untuk membandingkan lokasi memori dari dua objek, bukan sekadar membandingkan nilainya saja.
| Operator | Deskripsi | Contoh Ekspresi | Hasil (`x = [1, 2]`, `y = [1, 2]`, `z = x`) |
| :---: | :--- | :---: | :---: |
| `is` | Bernilai `True` jika kedua variabel merujuk pada objek yang sama di memori | `x is z` | `True` |
| `is not` | Bernilai `True` jika kedua variabel merujuk pada objek yang berbeda | `x is not y` | `True` |


### 2.1.6 Operator Perbandingan.
Operator perbandingan digunakan untuk membandingkan dua nilai. Hasil operasi perbandingan ini berupa nilai logika Boolean (True atau False).

| Operator | Deskripsi | Contoh | Hasil (jika `a = 5`, `b = 3`) |
| :---: | :--- | :---: | :---: |
| `==` | Sama dengan | `a == b` | `False` |
| `!=` | Tidak sama dengan | `a != b` | `True` |
| `>` | Lebih besar dari | `a > b` | `True` |
| `<` | Lebih kecil dari | `a < b` | `False` |
| `>=` | Lebih besar dari atau sama dengan | `a >= b` | `True` |
| `<=` | Lebih kecil dari atau sama dengan | `a <= b` | `False` |

### 2.2.1 Program "Aritmatika_2611532019"
Pada bagian ini, program digunakan untuk mengimplementasikan penggunaan operator aritmatika pada Python, seperti penjumlahan (+), pengurangan (-), pembagian (/), pembagian bulat (//), sisa bagi (%), dan perpangkatan (**).

```python
angka1_2019 = int(input("Input angka ke-1 : "))
angka2_2019 = int(input("Input angka ke-2 : "))

#Penjumlahan 
hasil_2019 = angka1_2019 + angka2_2019
print("\nOperator Penjumlahan")
print("Hasil = ",hasil_2019)

#Penjumlahan 
hasil_2019 = angka1_2019 - angka2_2019
print("\nOperator Pengurangan")
print("Hasil = ",hasil_2019)

# Pembagian, pembagian bulat, dan sisa bagi
if angka2_2019 != 0:
    hasil_2019 = angka1_2019 / angka2_2019
    print("\nOperator Pembagian")
    print("Hasil = ",hasil_2019)

    hasil_2019 = angka1_2019 // angka2_2019
    print("\nOperator Pembagian Bulat")
    print("Hasil = ",hasil_2019)

    hasil_2019 = angka1_2019 % angka2_2019
    print("\nOperator Sisa Bagi")
    print("Hasil = ",hasil_2019)
else:
    print("Angka kedua tidak boleh bernilai 0")

# Pangkat
hasil_2019 = angka1_2019**angka2_2019
print("\nOperator Perpangkatan Bulat")
print("Hasil = ",hasil_2019)
```

4. **Pembagian Bulat (`//`):** Digunakan untuk membagi `angka1_2019` dengan `angka2_2019` lalu membulatkan hasilnya ke bawah (*floor division*).
5. **Sisa Bagi (`%`):** Digunakan untuk mencari sisa hasil bagi (*modulus*) dari pembagian `angka1_2019` dengan `angka2_2019`.
6. **Perpangkatan (`**`):** Digunakan untuk menghitung hasil `angka1_2019` dipangkatkan dengan `angka2_2019`.

### 2.2.2 Program "Assignment_2611532019"
Pada bagian ini, program digunakan untuk mengimplementasikan penggunaan operator penugasan (assignment) pada Python, meliputi penugasan dasar (=) serta operator penugasan gabungan seperti penambahan (+=), pengurangan (-=), perkalian (*=), pembagian (/=), pembagian bulat (//=), sisa bagi (%=), dan perpangkatan (**=).

```python
angka1_2019 = int(input("Input angka ke-1 : "))
angka2_2019 = int(input("Input angka ke-2 : "))

print("\nNilai awal angka1 =",angka1_2019)
print("\nNilai awal angka2 =",angka2_2019)

#Assignment biasa
hasil = angka1_2019
print("\nAssignment biasa(=)")
print("Hasil =",hasil)

#Assignment penambahan
hasil = angka1_2019
hasil += angka2_2019
print("\nAssignment penambahan(+=)")
print("Hasil =",hasil)

#Assignment pengurangan
hasil = angka1_2019
hasil -= angka2_2019
print("\nAssignment pengurangan(-=)")
print("Hasil =",hasil)

#Assignment perkalian 
hasil = angka1_2019
hasil *= angka2_2019
print("\nAssignment perkalian(*=)")
print("Hasil =",hasil)

#Assignment pembagian, pembagian bulat, dan sisa bagi 
if angka2_2019 != 0:
    hasil = angka1_2019
    hasil /= angka2_2019
    print("\nAssignment pembagian(/=)")
    print("Hasil =",hasil)
    #Operator tambahan
    #Pembagian bulat
    hasil = angka1_2019
    hasil //= angka2_2019
    print("\nAssignment pembagian(//=)")
    print("Hasil =",hasil)
    #Sisa bagi
    hasil = angka1_2019
    hasil %= angka2_2019
    print("\nAssignment pembagian(%=)")
    print("Hasil =",hasil)
else:
    print("\nPembagian tidak dapat")
    print("Angka kedua tidak boleh bernilai 0.")

# Operator tambahan: assignment perpangkatan
hasil = angka1_2019
hasil **= angka2_2019
print("\nAssignment perpangkatan (**=)")
print("Hasil =",hasil)
```

**Penjelasan :**

1. **Assignment Biasa (`=`):** Digunakan untuk memasukkan nilai dari `angka1_2019` ke dalam variabel `hasil`.
2. **Assignment Penambahan (`+=`):** Digunakan untuk menambahkan nilai `hasil` dengan `angka2_2019` (`hasil = hasil + angka2_2019`).
3. **Assignment Pengurangan (`-=`):** Digunakan untuk mengurangkan nilai `hasil` dengan `angka2_2019` (`hasil = hasil - angka2_2019`).
4. **Assignment Perkalian (`*=`):** Digunakan untuk mengalikan nilai `hasil` dengan `angka2_2019` (`hasil = hasil * angka2_2019`).
5. **Assignment Pembagian (`/=`):** Digunakan untuk membagi nilai `hasil` dengan `angka2_2019` (`hasil = hasil / angka2_2019`).
6. **Assignment Pembagian Bulat (`//=`):** Digunakan untuk membagi nilai `hasil` dengan `angka2_2019` lalu membulatkannya ke bawah (`hasil = hasil // angka2_2019`).
7. **Assignment Sisa Bagi (`%=`):** Digunakan untuk menghitung sisa hasil bagi nilai `hasil` dengan `angka2_2019` (`hasil = hasil % angka2_2019`).
8. **Assignment Perpangkatan (`**=`):** Digunakan untuk memangkatkan nilai `hasil` dengan `angka2_2019` (`hasil = hasil ** angka2_2019`).

### 2.2.3 Program "Assignment_2611532019"
Pada bagian ini, program digunakan untuk mengimplementasikan penggunaan operator bitwise pada Python untuk melakukan operasi pada tingkat biner (bit), meliputi operasi AND (&), OR (|), XOR (^), NOT (~), serta pergeseran bit yaitu Shift Left (<<) dan Shift Right (>>), lengkap dengan penampil format biner menggunakan fungsi bin() dan format().

```python
print("\n===================")
print("OPERATOR BITWISE")
print("===================")

angka1_2019 = int(input("Masukkan angka bitwise-1: "))
angka2_2019 = int(input("Masukkan angka bitwise-2: "))

print("\nAngka dalam bentuk desimal dan biner")
print("angka1_2019 =", angka1_2019, "| biner", bin(angka1_2019))
print("angka2_2019 =", angka2_2019, "| biner", bin(angka2_2019))

# Bitwise AND
hasil_2019 = angka1_2019 & angka2_2019
print("\nBitwise AND (&)")
print(angka1_2019, "&", angka2_2019, "=", hasil_2019)
print("Biner hasil =", bin(hasil_2019))
print("Biner hasil (8 bit) =", format(hasil_2019, "08b"))

# Bitwise OR
hasil_2019 = angka1_2019 | angka2_2019
print("\nBitwise OR (|)")
print(angka1_2019, "|", angka2_2019, "=", hasil_2019)
print("Biner hasil =", bin(hasil_2019))
print("Biner hasil (8 bit) =", format(hasil_2019, "08b"))

# Bitwise XOR
hasil_2019 = angka1_2019 ^ angka2_2019
print("\nBitwise XOR (^)")
print(angka1_2019, "^", angka2_2019, "=", hasil_2019)
print("Biner hasil =", bin(hasil_2019))
print("Biner hasil (8 bit) =", format(hasil_2019, "08b"))

# Bitwise NOT
hasil_2019 = ~angka1_2019
print("\nBitwise NOT (~)")
print("~", angka1_2019, "=", hasil_2019)
print("Biner hasil =", bin(hasil_2019))
print("Biner hasil (8 bit) =", format(hasil_2019, "08b"))

# Bitwise geser kiri
jumlah_geser_2019 = int(input("\nMasukkan jumlah pergeseran bit: "))

hasil_2019 = angka1_2019 << jumlah_geser_2019
print("\nBitwise geser kiri (<<)")
print(angka1_2019, "<<", jumlah_geser_2019, "=", hasil_2019)
print("Biner hasil =", bin(hasil_2019))
print("Biner hasil (8 bit) =", format(hasil_2019, "08b"))

# Bitwise geser kanan
hasil_2019 = angka1_2019 >> jumlah_geser_2019
print("\nBitwise geser kanan (>>)")
print(angka1_2019, ">>", jumlah_geser_2019, "=", hasil_2019)
print("Biner hasil =", bin(hasil_2019))
print("Biner hasil (8 bit) =", format(hasil_2019, "08b"))
```

**Penjelasan :**

1. **Konversi Biner (`bin()` dan `format()`):** Fungsi `bin()` digunakan untuk menampilkan representasi biner dari suatu angka, sedangkan `format(..., "08b")` digunakan untuk menampilkan bentuk biner berformat 8-bit.
2. **Bitwise AND (`&`):** Membandingkan bit demi bit dari dua angka dan menghasilkan nilai `1` hanya jika kedua bit sama-sama bernilai `1`.
3. **Bitwise OR (`|`):** Membandingkan bit demi bit dan menghasilkan nilai `1` jika salah satu atau kedua bit bernilai `1`.
4. **Bitwise XOR (`^`):** Membandingkan bit demi bit dan menghasilkan nilai `1` hanya jika kedua bit memiliki nilai yang berbeda.
5. **Bitwise NOT (`~`):** Membalikkan nilai setiap bit (operasi *bitwise complement*) dari `angka1_2019` (`0` menjadi `1` dan `1` menjadi `0`).
6. **Bitwise Geser Kiri (`<<`):** Menggeser bit dari `angka1_2019` ke arah kiri sebanyak nilai `jumlah_geser_2019` (secara matematis sama dengan mengalikan angka dengan $2^n$).
7. **Bitwise Geser Kanan (`>>`):** Menggeser bit dari `angka1_2019` ke arah kanan sebanyak nilai `jumlah_geser_2019` (secara matematis sama dengan membagi angka dengan $2^n$ lalu membulatkannya ke bawah).

### 2.2.4 Program "Logika_2611532019"
Pada bagian ini, program digunakan untuk mengimplementasikan penggunaan operator logika pada Python untuk mengevaluasi ekspresi nilai kebenaran (boolean), meliputi Konjungsi (and), Disjungsi (or), Negasi (not), serta Disjungsi Eksklusif (XOR) yang dibentuk menggunakan operator perbandingan (!=).

```python
a1 = input("Input nilai boolean-1 (True/false): ").strip().lower()== "true"
a2 = input("Input nilai boolean-2 (True/false): ").strip().lower()== "true"

print("\nA1", a1)
print("\nA2", a2)

# Konjungsi : bernilai True jika keduanya True
hasil = a1 and a2
print("\nKonjungsi(AND)")
print("A1 and A2 =", hasil)

# Disjungsi : bernilai True jika salah satunya True
hasil = a1 or a2
print("\nDisjungsi(OR)")
print("A1 or A2 =", hasil)

# Negasi A1 : membalik nilai A1
hasil = not a1
print("\nNegasi A1(AND)")
print("not A1 =", hasil)

# Negasi A2 : membalik nilai A2
hasil = not a2
print("\nNegasi A2(NOT)")
print("not A2 =", hasil)

#XOR: bernilai True jika kedua nilai berbeda
hasil = a1 != a2
print("\nDisjungsi Ekslusif (XOR)")
```

**Penjelasan :**

1. **Input Boolean (`.strip().lower() == "true"`):** Menerima masukan teks dari pengguna, menghapus spasi di awal/akhir dengan `.strip()`, mengubah huruf menjadi kecil dengan `.lower()`, lalu membandingkannya dengan kata `"true"` untuk menghasilkan tipe data *boolean* (`True` atau `False`).
2. **Konjungsi / AND (`and`):** Operator yang menghasilkan nilai `True` hanya jika kedua nilai *boolean* (`a1` dan `a2`) sama-sama bernilai `True`.
3. **Disjungsi / OR (`or`):** Operator yang menghasilkan nilai `True` jika salah satu atau kedua nilai *boolean* (`a1` atau `a2`) bernilai `True`.
4. **Negasi / NOT (`not`):** Operator yang membalikkan nilai *boolean* (jika nilai awal `True` akan menjadi `False`, dan sebaliknya).
5. **Disjungsi Eksklusif / XOR (`!=`):** Menggunakan operator perbandingan `!=` untuk mensimulasikan logika XOR, yaitu menghasilkan nilai `True` hanya jika kedua nilai *boolean* (`a1` dan `a2`) memiliki nilai yang berbeda.

### 2.2.5 Program "Nama_lainnya_2611532019"
Pada bagian ini, program digunakan untuk mengimplementasikan penggunaan operator keanggotaan (membership) meliputi IN (in) dan NOT IN (not in) untuk mengecek keberadaan suatu elemen dalam list, serta operator identitas (identity) meliputi IS (is) dan IS NOT (is not) untuk membandingkan referensi alamat memori dari objek yang dibuat secara langsung maupun melalui metode pengkopian (.copy()).

```python
print("=========================")
print("1. OPERATOR KEANGGOTAAN")
print("=========================")

#Input beberarpa data yang ingin dipisahkan dengan koma
input_data_2019 = print("Masukkan beberapa angka, pisahkan dengan koma : ")

# Mengubah input menjadi list integer
data_2019 = [int(angka.strip()) for angka in input_data_2019(",")]

nilai_dicari_2019 = int(input("Masukkan angka yang ingin dicari"))

# Operator in
hasil_2019 = nilai_dicari_2019 in data_2019
print("\nOperator keanggotaan IN")
print(nilai_dicari_2019,"in",data_2019,"=",hasil_2019)

#Operator not in
hasil_2019 = nilai_dicari_2019  not in data_2019
print("\nOperator keanggotaan not IN")
print(nilai_dicari_2019,"not in",data_2019,"=",hasil_2019)

print("=========================")
print("2. OPERATOR IDENTITAS")
print("=========================")

# objek1_2019 menggunakan list dari input pengguna
objek1_2019 = data_2019

# objek2_2019 merujuk pada objek yang sama dengan objek1_2019
objek2_2019 = objek1_2019

# objek3_2019 memiliki isi sama, tetapi merupakan objek baru
objek3_2019 = data_2019.copy()

print("objek1_2019 =", objek1_2019)
print("objek2_2019 =", objek2_2019)
print("objek3_2019 =", objek3_2019)

# Operator is
hasil_2019 = objek1_2019 is objek2_2019
print("\nOperator identitas IS")
print("objek1_2019 is objek2_2019 =", hasil_2019)

# Operator is not
hasil_2019 = objek1_2019 is not objek3_2019
print("\nOperator identitas IS NOT")
print("objek1_2019 is not objek3_2019 =", hasil_2019)

# Membandingkan identitas dan nilai
print("\nPerbandingan identitas dan nilai")
print("objek1_2019 is objek3_2019 =", objek1_2019 is objek3_2019)
print("objek1_2019 == objek3_2019 =", objek1_2019 == objek3_2019)
```

**Penjelasan :**

1. **Pengolahan Input List (`[int(...) for ...]`):** Mengonversi deretan angka yang diinputkan pengguna dan dipisahkan oleh tanda koma menjadi *list* bernilai bilangan bulat (*integer*).
2. **Operator Keanggotaan IN (`in`):** Memeriksa apakah `nilai_dicari_2019` terdapat di dalam `data_2019` dan menghasilkan nilai `True` jika angka tersebut ditemukan.
3. **Operator Keanggotaan NOT IN (`not in`):** Memeriksa apakah `nilai_dicari_2019` tidak terdapat di dalam `data_2019` dan menghasilkan nilai `True` jika angka tersebut memang tidak ada dalam *list*.
4. **Inisialisasi Referensi Objek Memori:**
   * `objek2_2019 = objek1_2019`: Membuat variabel `objek2_2019` yang merujuk pada alamat memori yang **sama persis** dengan `objek1_2019`.
   * `objek3_2019 = data_2019.copy()`: Duplikasi isi *list* menggunakan `.copy()`, sehingga `objek3_2019` memiliki nilai isi yang sama tetapi disimpan pada **objek/alamat memori yang terpisah**.
5. **Operator Identitas IS (`is`):** Mengevaluasi apakah dua variabel menunjuk ke lokasi memori yang sama. Menghasilkan `True` pada `objek1_2019 is objek2_2019`.
6. **Operator Identitas IS NOT (`is not`):** Mengevaluasi apakah dua variabel menunjuk ke lokasi memori yang berbeda. Menghasilkan `True` pada `objek1_2019 is not objek3_2019`.
7. **Perbedaan Operator `is` dan `==`:** Menunjukkan bahwa `objek1_2019 is objek3_2019` bernilai `False` (karena identitas/lokasi memori berbeda), sedangkan `objek1_2019 == objek3_2019` bernilai `True` (karena nilai isinya sama).

### 2.2.6 Program "Perbandingan_2611532019"
Pada bagian ini, program digunakan untuk mengimplementasikan penggunaan operator perbandingan (relational) pada Python untuk membandingkan dua nilai, meliputi Lebih Besar (>), Lebih Kecil (<), Lebih Besar dari atau Sama Dengan (>=), Lebih Kecil dari atau Sama Dengan (<=), Sama Dengan (==), Tidak Sama Dengan (!=), serta penerapan Perbandingan Berantai (chained comparison).

```python
angka1_2019 = int(input("Input angka ke-1 : "))
angka2_2019 = int(input("Input angka ke-2 : "))

#Lebih besar dari 
hasil_2019 = angka1_2019 > angka2_2019
print("\nOperator Lebih Besar")
print("Hasil = ",hasil_2019)

#Lebih kecil dari 
hasil_2019 = angka1_2019 < angka2_2019
print("\nLebih kecil dari ")
print("Hasil = ",hasil_2019)

#Lebih besar dari atau sama dengan 
hasil_2019 = angka1_2019 >= angka2_2019
print("\nOperator Lebih besar dari atau sama dengan")
print("angka1 >= angka2 = ",hasil_2019)

#Lebih kecil dari atau sama dengan
hasil_2019 = angka1_2019 <= angka2_2019
print("\nOperator Lebih kecil dari atau sama dengan")
print("angka1 <= angka2 = ",hasil_2019)

#Sama dengan 
hasil_2019 = angka1_2019 == angka2_2019
print("\nOperator Sama dengan")
print("angka1 == angka2 = ",hasil_2019)

#Tidak sama dengan 
hasil_2019 = angka1_2019 != angka2_2019
print("\nOperator Tidak sama dengan")
print("Hasil = ",hasil_2019)


#Tambahan: Perbandingan berantai dalam python
hasil_2019 = 0 < angka1_2019 < 100
print("\nOperator Perbandingan berantai")
print("0 < angka1 < 100 ",hasil_2019)

hasil_2019 = 0 < angka2_2019 < 100
print("0 < angka2 < 100",hasil_2019)
```

**Penjelasan :**

1. **Operator Lebih Besar (`>`):** Memeriksa apakah nilai `angka1_2019` lebih besar daripada `angka2_2019`. Hasilnya bernilai `True` jika kondisi tersebut terpenuhi.
2. **Operator Lebih Kecil (`<`):** Memeriksa apakah nilai `angka1_2019` lebih kecil daripada `angka2_2019`.
3. **Operator Lebih Besar dari atau Sama Dengan (`>=`):** Memeriksa apakah nilai `angka1_2019` bernilai lebih besar dari atau minimal sama dengan `angka2_2019`.
4. **Operator Lebih Kecil dari atau Sama Dengan (`<=`):** Memeriksa apakah nilai `angka1_2019` bernilai lebih kecil dari atau maksimal sama dengan `angka2_2019`.
5. **Operator Sama Dengan (`==`):** Memeriksa apakah nilai `angka1_2019` memiliki nilai yang persis sama dengan `angka2_2019`.
6. **Operator Tidak Sama Dengan (`!=`):** Memeriksa apakah nilai `angka1_2019` berbeda atau tidak sama dengan nilai `angka2_2019`.
7. **Perbandingan Berantai (`0 < angka1_2019 < 100`):** Fitur khusus Python yang memungkinkan pengecekan dua kondisi sekaligus dalam satu baris (memeriksa apakah nilai variabel berada dalam rentang antara `0` dan `100`).

```python
# ATURAN PENAMAAN VARIABEL & DICTIONARY BARANG

produk_2019 = {
    "1": {"nama": "Sabun", "harga": 5000},
    "2": {"nama": "Beras 5Kg", "harga": 70000},
    "3": {"nama": "Beras 10Kg", "harga": 130000},
    "4": {"nama": "Aqua 1L", "harga": 6000},
    "5": {"nama": "Milo Kaleng", "harga": 130000},
    "6": {"nama": "Indomie Goreng", "harga": 3500},
    "7": {"nama": "Indomilk Sachet", "harga": 130000},
}

# Ketentuan Tambahan
membership_fee_2019 = 30000
biaya_tambahan_2019 = 0

print("=== SISTEM TRANSAKSI TOKO ===\n")

# Data Pelanggan
nama_2019 = input("Masukkan Nama Anda : ")
input_status_2019 = input(f"Selamat datang {nama_2019}\nApakah anda adalah member di toko kami? (member/nonmember) : ").strip().lower()

is_member_2019 = False

# Menentukan Member / Non-Member
if input_status_2019 in ["member", "ya", "y"]:
    is_member_2019 = True
    print("Wah ada member, nikmatin diskonnya ya!\n")
elif input_status_2019 in ["nonmember", "tidak", "n"]:
    minat_2019 = input("Apakah berminat berlangganan membership? (30Rb/bulan, diskon 10%): (Ya/Tidak) : ").strip().lower()
    if minat_2019 in ["ya", "iya", "y"]:
        is_member_2019 = True
        biaya_tambahan_2019 += membership_fee_2019  # Operator Augmented Assignment (+=)
        print("Selamat anda sudah menjadi member di toko kami!\n")
    else:
        is_member_2019 = False

# LOOPING PEMBELIAN BARANG DARI DICTIONARY

total_belanja_2019 = 0
total_item_2019 = 0
daftar_pembelian_2019 = []

menu_2019 = True
while menu_2019:
    print(f"\n{nama_2019}, mau beli apa? Ini daftarnya:")
    for nomor_2019, barang_2019 in produk_2019.items():
        print(f'{nomor_2019}. {barang_2019["nama"]} - Rp{barang_2019["harga"]}')

    pilihan_2019 = input("\nMasukkan nomor barang: ")
    if pilihan_2019 in produk_2019:
        jumlah_2019 = int(input("Masukkan jumlah barang: "))
        item_2019 = produk_2019[pilihan_2019]
        
        subtotal_2019 = item_2019["harga"] * jumlah_2019
        total_belanja_2019 += subtotal_2019
        total_item_2019 += jumlah_2019
        daftar_pembelian_2019.append(f"{item_2019['nama']} (x{jumlah_2019})")
        print(f"-> {item_2019['nama']} dimasukkan ke keranjang.")
    else:
        print("Nomor barang tidak valid!")

    tanya_2019 = input("\nApakah ingin membeli barang lain? (y/n): ").strip().lower()
    if tanya_2019 != 'y':
        menu_2019 = False

# Input Kode Promo
kode_promo_2019 = input("\nMasukkan Kode Promo : ").strip().upper()


# OPERATOR KEANGGOTAAN (MEMBERSHIP)

daftar_promo_2019 = ["HEMAT10", "HEMAT20", "GRATISONGKIR"]
promo_tersedia_2019 = kode_promo_2019 in daftar_promo_2019

# OPERATOR PERBANDINGAN & OPERATOR LOGIKA

syarat_belanja_2019 = total_belanja_2019 >= 200000
syarat_barang_2019 = total_item_2019 >= 3

# Evaluasi Logika
dapat_diskon_member_2019 = is_member_2019 and syarat_belanja_2019
dapat_promo_2019 = promo_tersedia_2019 or (syarat_barang_2019 and not is_member_2019)

# OPERATOR ARITMATIKA & OPERATOR PENUGASAN
persen_diskon_2019 = 0.0

if dapat_diskon_member_2019:
    persen_diskon_2019 += 0.10

if dapat_promo_2019 and kode_promo_2019 == "HEMAT10":
    persen_diskon_2019 += 0.10

besarnya_diskon_2019 = total_belanja_2019 * persen_diskon_2019
total_pembayaran_2019 = (total_belanja_2019 - besarnya_diskon_2019) + biaya_tambahan_2019
rata_rata_harga_2019 = total_belanja_2019 / total_item_2019 if total_item_2019 > 0 else 0

# OPERATOR IDENTITAS (IDENTITY: is, is not)

ref_member_2019 = True
cek_identitas_2019 = is_member_2019 is ref_member_2019

# OPERATOR BITWISE

BIT_MEMBER_2019     = 0b0001
BIT_TOTAL_200K_2019 = 0b0010
BIT_BARANG_3_2019   = 0b0100
BIT_PROMO_2019      = 0b1000

kode_status_2019 = 0b0000

# Bitwise OR (|)
if is_member_2019:
    kode_status_2019 |= BIT_MEMBER_2019
if syarat_belanja_2019:
    kode_status_2019 |= BIT_TOTAL_200K_2019
if syarat_barang_2019:
    kode_status_2019 |= BIT_BARANG_3_2019
if promo_tersedia_2019:
    kode_status_2019 |= BIT_PROMO_2019

# Bitwise AND (&)
cek_member_bit_2019 = kode_status_2019 & BIT_MEMBER_2019
cek_promo_bit_2019 = kode_status_2019 & BIT_PROMO_2019

# Bitwise XOR (^)
kode_referensi_2019 = 0b1011
hasil_xor_2019 = kode_status_2019 ^ kode_referensi_2019

# Bitwise Shift (<<)
hasil_shift_2019 = kode_status_2019 << 1



# OUTPUT HASIL SESUAI HAK AKSES & MODUL

print("\n=== DATA TRANSAKSI ===")
print(f"Nama Pelanggan        : {nama_2019}")
print(f"Status Pelanggan      : {'member' if is_member_2019 else 'nonmember'}")
print(f"Total Belanja         : Rp{int(total_belanja_2019)}")
print(f"Jumlah Barang         : {total_item_2019}")
print(f"Kode Promo            : {kode_promo_2019}")

print("\n=== HASIL VALIDASI ===")
print(f"Belanja >= Rp200000        : {syarat_belanja_2019}")
print(f"Jumlah Barang >= 3         : {syarat_barang_2019}")
print(f"Status Member              : {is_member_2019}")
print(f"Kode Promo Tersedia        : {promo_tersedia_2019}")
print(f"Mendapatkan Diskon         : {dapat_diskon_member_2019}")
print(f"Mendapatkan Promo          : {dapat_promo_2019}")

print("\n=== HASIL PERHITUNGAN ===")
print(f"Diskon                     : Rp{int(besarnya_diskon_2019)}")
print(f"Total Pembayaran           : Rp{int(total_pembayaran_2019)}")
print(f"Rata-rata Harga Barang     : Rp{rata_rata_harga_2019:.0f}")

print("\n=== HAK AKSES PELANGGAN ===")
print(f"Kode Hak Akses             : {bin(kode_status_2019)[2:].zfill(4)}")
print(f"Member Access              : {bool(cek_member_bit_2019)}")
print(f"Promo Access               : {bool(cek_promo_bit_2019)}")
print(f"Free Shipping Access       : {bool(kode_status_2019 & BIT_TOTAL_200K_2019)}")

print("\n=== OPERASI BITWISE ===")
print("=== Kode Status Transaksi ===")
print("0001 | 0010 | 0100 | 1000")
print(f"Kode Biner    : {bin(kode_status_2019)[2:].zfill(4)}")
print(f"Kode Desimal  : {kode_status_2019}")

print("\n=== Pemeriksaan Status ===")
print("Cek Member")
print(f"{bin(kode_status_2019)[2:].zfill(4)} & 0001")
print(f"Hasil Biner   : {bin(cek_member_bit_2019)[2:].zfill(4)}")
print(f"Hasil Desimal : {cek_member_bit_2019}")

print("\nCek Promo")
print(f"{bin(kode_status_2019)[2:].zfill(4)} & 1000")
print(f"Hasil Biner   : {bin(cek_promo_bit_2019)[2:].zfill(4)}")
print(f"Hasil Desimal : {cek_promo_bit_2019}")

print("\n=== Perbandingan Status ===")
print(f"Kode Transaksi : {bin(kode_status_2019)[2:].zfill(4)}")
print(f"Kode Referensi : {bin(kode_referensi_2019)[2:].zfill(4)}")
print(f"{bin(kode_status_2019)[2:].zfill(4)} ^ {bin(kode_referensi_2019)[2:].zfill(4)}")
print(f"Hasil Biner   : {bin(hasil_xor_2019)[2:].zfill(4)}")
print(f"Hasil Desimal : {hasil_xor_2019}")

print("\n=== Shift ===")
print(f"{bin(kode_status_2019)[2:].zfill(4)} << 1")
print(f"Hasil Biner   : {bin(hasil_shift_2019)[2:]}")
print(f"Hasil Desimal : {hasil_shift_2019}")

print("\n=== SELESAI ===")










#2611532019
```
