## 2.1 Dasar Teori

Program pada praktikum ini menggunakan beberapa struktur percabangan untuk menentukan keputusan berdasarkan data pengguna. Percabangan membuat alur program dapat menyesuaikan hasil dengan kondisi yang terpenuhi.

### 2.1.1 Percabangan if

if digunakan untuk menjalankan blok kode apabila suatu kondisi bernilai True.

```python
if tiket_2019 > 0:
    print("Wahana dapat dipilih")
```

### 2.1.2 Percabangan if-elif-else

Struktur if-elif-else digunakan ketika terdapat beberapa kemungkinan kondisi. Python memeriksa kondisi dari atas ke bawah dan menjalankan blok pertama yang bernilai benar.

```python
if tiket_2019 > 0:
    print("Tiket valid")
elif tiket_2019 == 0:
    print("Tidak memiliki tiket")
else:
    print("Jumlah tiket tidak valid")
```

### 2.1.3 Percabangan Bersarang

Percabangan bersarang adalah percabangan yang berada di dalam percabangan lain. Struktur ini digunakan ketika keputusan kedua hanya perlu diperiksa setelah kondisi pertama terpenuhi.

```python
if paket_wahana_2019 in ["1", "5"]:
    if umur_2019 >= 14 and tinggi_2019 > 150:
        print("Persyaratan terpenuhi")
```

### 2.1.4 Percabangan match-case

match-case digunakan untuk mencocokkan sebuah nilai dengan beberapa pola pilihan. Pada program ini, match-case digunakan untuk menentukan nama dan harga wahana berdasarkan nomor yang dipilih pengguna.

```python
match paket_wahana_2019:
    case "1":
        harga_satuan_2019 = 50000
    case "2":
        harga_satuan_2019 = 30000
    case _:
        print("Wahana tidak tersedia")
```

### 2.1.5 Operator Logika pada Percabangan

Operator and, or, dan not digunakan untuk menggabungkan atau membalik kondisi.

- and bernilai True jika semua kondisi benar.
- or bernilai True jika salah satu kondisi benar.
- not membalik nilai kebenaran suatu kondisi.

Contoh:

```python
if umur_2019 >= 17 and sim_2019 == "y":
    print("Boleh menggunakan wahana kendaraan")
```

### 2.1.6 Kondisi dan Validasi Input

Kondisi digunakan untuk memeriksa apakah data pengguna sesuai dengan aturan wahana. Contohnya, tinggi badan harus berada pada rentang tertentu, umur harus memenuhi batas minimum, dan jumlah tiket harus lebih dari nol.

Validasi input membantu mencegah program memproses data yang tidak sesuai.

### 2.2.1 Program if1_2019.py

Program ini menggunakan percabangan if untuk menampilkan keterangan lulus apabila IPK lebih besar dari 2,75. Jika kondisi tidak terpenuhi, program hanya menampilkan pesan selesai.

````python
ipk_2019 = float(input("input ipk anda : "))

if ipk_2019 > 2.75:
    print("Anda Lulus Dengan Memuaskan dengan IPK" + str(ipk_2019))

print("Program Selesei")
````

### 2.2.2 Program if2_2019.py

Program ini menggunakan percabangan if-else. Jika IPK lebih besar dari 2,75, pengguna dinyatakan lulus dengan memuaskan. Selain itu, program menampilkan pesan bahwa pengguna tidak lulus.

````python
ipk_2019 = float(input("input ipk anda : "))

if ipk_2019 > 2.75:
    print("Anda Lulus Dengan Memuaskan dengan IPK" + str(ipk_2019))
else:
    print("Anda tidak lulus")

print("Program Selesei")
````

Penjelasan:

1. Program menerima IPK dalam bentuk float.
2. Kondisi ipk_2019 > 2.75 diperiksa oleh percabangan if.
3. Blok else dijalankan apabila kondisi pada if bernilai False.
4. Dengan demikian, setiap nilai IPK memperoleh keluaran sesuai kondisi yang terpenuhi.

### 2.2.4 Program if_elif_else1_2019.py

Program ini menggunakan if-elif-else dan operator logika and untuk menentukan kelayakan pengguna membawa motor berdasarkan umur dan kepemilikan SIM.

````python
umur_2019 = int(input("Input umur anda : "))
sim_2019 = input("Apakah anda sudah punya SIM c (y/t) : ")[0].lower()

if umur_2019 >= 17 and sim_2019 == "y":
    print("Anda sudah dewasa dan boleh bawa motor")
elif umur_2019 >= 17 and sim_2019 != "y":
    print("Anda sudah dewasa tetapi tidak boleh bawa motor")
elif umur_2019 < 17 and sim_2019 == "y":
    print("Anda belum cukup umur punya SIM")
else:
    print("Anda belum cukup umur bawa motor")

print("Program selesei")
````

Penjelasan:

1. Program membaca umur dan status kepemilikan SIM.
2. Operator and mengharuskan dua kondisi terpenuhi sekaligus.
3. if-elif-else memisahkan empat kemungkinan kombinasi umur dan SIM.
4. Percabangan terakhir digunakan untuk pengguna yang belum cukup umur dan tidak memiliki SIM.

### 2.2.5 Program multi_if1_2019.py

Program ini menggunakan beberapa if terpisah untuk memeriksa syarat diskon. Setiap kondisi diperiksa secara independen sehingga diskon dapat diakumulasikan.

````python
total_belanja_2019 = float(input("Masukkan total belanja (rp) :"))

input_member_2019 = input("Apakah anda member? (y/t) : ")
is_member_2019 = input_member_2019 in ["y", "ya", "iya"]

input_promo_2019 = input("Apakah kode promo valid (y/n) ").strip().lower()
kode_promo_valid_2019 = input_promo_2019 in ["ya", "iya", "y"]

total_diskon_persen_2019 = 0

if total_belanja_2019 > 1000000:
    total_diskon_persen_2019 += 10

if is_member_2019:
    total_diskon_persen_2019 += 5

if kode_promo_valid_2019:
    total_diskon_persen_2019 += 15

nominal_diskon_2019 = total_belanja_2019 * (total_diskon_persen_2019 / 100)
total_bayar_2019 = total_belanja_2019 - nominal_diskon_2019

print("\n===Rincian Pembayaran===")
print(f"Total diskon : {total_diskon_persen_2019}% (Rp {nominal_diskon_2019:,.0f})")
print(f"Total Bayar  : Rp{total_bayar_2019}")
````

Penjelasan:

1. Program menerima total belanja, status member, dan validitas kode promo.
2. Setiap if berdiri sendiri sehingga lebih dari satu diskon dapat diberikan.
3. Operator += menambahkan persentase diskon yang diperoleh.
4. Nominal diskon dihitung dari total belanja dan persentase diskon.

### 2.2.6 Program multi_if2_2019.py

Program ini menggunakan beberapa percabangan if untuk menentukan izin membawa motor. Berbeda dengan if-elif-else, setiap kondisi diperiksa secara terpisah.

````python
umur_2019 = int(input("Input umur anda : "))
sim_2019 = input("Apakah anda sudah punya SIM c (y/t) : ")[0].lower()

if umur_2019 >= 17 and sim_2019 == "y":
    print("Anda sudah dewasa dan boleh bawa motor")

if umur_2019 >= 17 and sim_2019 != "y":
    print("Anda sudah dewasa tetapi tidak boleh bawa motor")

if umur_2019 < 17 and sim_2019 == "y":
    print("Anda belum cukup umur punya SIM")

if umur_2019 < 17 and sim_2019 != "y":
    print("Anda belum cukup umur bawa motor")

print("Program selesei")
````

Penjelasan:

1. Program memeriksa empat kondisi menggunakan empat pernyataan if terpisah.
2. Kondisi pertama memeriksa pengguna dewasa yang memiliki SIM.
3. Kondisi kedua memeriksa pengguna dewasa yang tidak memiliki SIM.
4. Kondisi ketiga dan keempat memeriksa pengguna yang belum dewasa.
5. Karena setiap if berdiri sendiri, seluruh kondisi tetap dievaluasi satu per satu.

### 2.2.7 Program match_case1.py

Program ini menggunakan match-case untuk mencocokkan nomor bulan dengan nama bulan. case _ digunakan sebagai kondisi bawaan apabila angka yang dimasukkan berada di luar rentang 1 sampai 12.

Kode Program:

File match_case1.py menggunakan struktur match-case untuk memilih keluaran berdasarkan angka bulan.

````python
bulan_2019 = int(input("Masukkan angka bulan (1-12) : "))

match bulan_2019:
    case 1:
        print("Januari")
    case 2:
        print("Februari")
    case 3:
        print("Maret")
    case 4:
        print("April")
    case 5:
        print("Mei")
    case 6:
        print("Juni")
    case 7:
        print("Juli")
    case 8:
        print("Agustus")
    case 9:
        print("September")
    case 10:
        print("Oktober")
    case 11:
        print("November")
    case 12:
        print("Desember")
    case _:
        print("Angka tidak valid")
````

### 2.3.2 Kode Jawaban

Program berikut merupakan implementasi tugas praktikum Pertemuan 4.

````python
#Input data pengguna
print("====SISTEM LOKER WAHANA PRANSMART====")
print("Selamat datang di Wahana Pransmart")
nama_2019 = input("Masukkan nama anda : ")
umur_2019 = int(input("Masukkan umur anda : "))
sim_2019 = input("Apakah anda sudah memiliki Sim? (y/n) : ").strip().lower()
tiket_2019 = int(input("Berapa tiket anda? : "))
member_2019 = input("Apakah anda member? (y/n) : ").strip().lower()

#Untuk check apakah tiket valid atau tidak
if tiket_2019 > 0:
    print("Wahana apa yang ingin anda kunjungi?")
elif tiket_2019 == 0:
    print("Anda tidak memiliki tiket\nSilahkan beli tiket terlebih dahulu")
else :
    print("Masukkan angka yang valid (positif)")

#Wahana yang tersedia
print("\n====DAFTAR WAHANA====\n1. Roller Coaster\n2. Rumah Hantu\n3. Waterboom\n4. Bumper Car\n5. Bianglala\n6. Motor Cross\n7. All access pass (semua wahana)")
paket_wahana_2019 = input("Masukkan nomor wahana yang ingin anda kunjungi : ")

match paket_wahana_2019:
    case "1":
        harga_satuan_2019 = 50000
        print("Anda memilih wahana Roller Coaster | Harga Rp. 50.000")
    case "2":
        harga_satuan_2019 = 30000
        print("Anda memilih wahana Rumah Hantu | Harga Rp. 30.000")
    case "3":
        harga_satuan_2019 = 40000
        print("Anda memilih wahana Waterboom | Harga Rp. 40.000")
    case "4":
        harga_satuan_2019 = 20000
        print("Anda memilih wahana Bumper Car | Harga Rp. 20.000")
    case "5":
        harga_satuan_2019 = 25000
        print("Anda memilih wahana Bianglala | Harga Rp. 25.000")
    case "6":
        harga_satuan_2019 = 50000
        print("Anda memilih wahana Motor Cross | Harga Rp. 50.000")
    case "7":
        harga_satuan_2019 = 150000
        print("Anda memilih All access pass | Harga Rp. 150.000")
    case _:
        print("Wahana yang anda pilih tidak tersedia")

print("\n====KETENTUAN WAHANA====")
#check tinggi badan untuk roller coaster dan bianglala
if paket_wahana_2019 in ["1", "5"]:
    tinggi_2019 = int(input("Masukkan tinggi badan anda (cm) : "))
    if umur_2019 >= 14 and tinggi_2019 > 150 and tinggi_2019 < 200:
        print("Tinggi badan anda memenuhi syarat untuk wahana ini")
    elif umur_2019 >= 14 and tinggi_2019 <= 150:
        print("Tinggi badan anda terlalu pendek untuk masuk ke wahana ini")
    elif umur_2019 >= 14 and tinggi_2019 >= 200:
        print("Tinggi badan anda terlalu tinggi untuk masuk ke wahana ini")
    elif umur_2019 < 14 and tinggi_2019 > 150 and tinggi_2019 < 200:
        print("Umur anda belum cukup untuk masuk ke wahana ini")
    else:
        print("Umur dan tinggi badan anda belum memenuhi syarat untuk masuk ke wahana ini")

#check sim untuk bumper car, motor cross, dan all access pass
if paket_wahana_2019 in ["4","6","7"]:
    if umur_2019 >= 17 and sim_2019 == "y":
        print("Anda sudah dewasa dan memiliki sim, anda boleh mengendarai motor dan mobil di wahana ini")
    elif umur_2019 >= 17 and sim_2019 != "y":
        print("Anda sudah dewasa tetapi tidak memiliki SIM , anda boleh mengendarai motor dan mobil di wahana ini asalkan dengan pendamping/pengawas")
    elif umur_2019 < 17 and sim_2019 == "y":
        print("Anda belum cukup umur untuk memiliki SIM , anda tidak boleh mengendarai motor dan mobil di wahana ini karena identitas anda tidak valid")
    else:
        print("Anda belum cukup umur dan tidak memiliki SIM , anda tidak boleh mengendarai motor dan mobil di wahana ini dimohon untuk memilih wahana lain")
        print("wahana yang dapat anda pilih adalah wahana Roller Coaster, Rumah Hantu, Waterboom, dan Bianglala")

#check umur untuk rumah hantu dan waterboom
if paket_wahana_2019 in ["2","3"]:
    print("Wahana ini memiliki batasan umur minimal 10 tahun")
    if int(umur_2019) >= 10:
        print("Umur anda sudah cukup untuk masuk ke wahana ini")
    else:
        print("Umur anda belum cukup untuk masuk ke wahana ini")

subtotal_2019 = harga_satuan_2019 * tiket_2019

#ketentuan promo
promo_2019 = ["weekend","promo11.11"] #kode promo yang berlaku
print("\n====KETENTUAN PROMO====")
print("1. Weekend       : Diskon 15% untuk pembelian tiket di hari sabtu dan minggu")
print("2. Promo11.11    : Diskon 15% untuk pembelian tiket di tanggal 11 November")
kode_promo_2019 = input("Masukkan kode promo anda (jika ada) : ").strip().lower()

if kode_promo_2019 in promo_2019:
    print("Kode promo : " + kode_promo_2019 + " | Kode promo anda valid")
    kode_promo_2019 = True
else :
    print("Kode promo anda tidak valid")
    kode_promo_2019 = False
    
#pajak yang berlaku
pajak_2019 = 0.015 

#diskon yang berlaku
diskon_2019 = 0

if subtotal_2019 >= 200000:
    diskon_2019 =+ 0.1

if member_2019 == ["y","iya","ya"]:
    diskon_2019 += 0.05

if tiket_2019 >= 5:
    diskon_2019 += 0.05

if kode_promo_2019 == True:
    diskon_2019 += 0.15

#nominal total
total_diskon_2019 = subtotal_2019 * diskon_2019
pajak_2019 = subtotal_2019 * pajak_2019
total_2019 = subtotal_2019 - total_diskon_2019 + pajak_2019

print("\n====Rincian Pembayaran====")
print(f"Subtotal belanja    : Rp.{subtotal_2019:,}")
print(f"Diskon              : {diskon_2019:.0%} (Rp.{subtotal_2019 * diskon_2019:,})")
print(f"Pajak PPN           : Rp.{pajak_2019:,.0f}")
print(f"Total               : Rp.{subtotal_2019:,} - Rp.{total_diskon_2019:,} + Rp.{pajak_2019:,}\n                    : Rp.{total_2019:,.0f}")
if total_2019 > 300000 :
    print(f"Catatan Layanan     : Terima kasih {nama_2019} telah berkunjung ke Wahana Pransmart, anda mendapatkan souvenir dari kami karena total belanja anda melebihi Rp.300.000")
else :  
    print(f"Catatan Layanan     : Terima kasih {nama_2019} telah berkunjung ke Wahana Pransmart")
print("Program Selesei")
````
