## 2.1 Dasar Teori
Praktikum pekan ke-2 ini berfokus pada analisis variabel dan pemrosesan berbagai jenis tipe data pada Python. Beberapa landasan teori utama yang digunakan meliputi:

1. **Variabel dan Aturan Penulisan**.
   - **Pengertian Variabel**: Tempat menyimpan data/nilai di memori.
   - **Aturan Penulisan (NIM: 2019)**:
     - Semua nama variabel wajib diakhiri dengan `_2019` (contoh: `nama_2019`, `total_belanja_2019`, `kode_status_2019`).
     - Bersifat *case-sensitive* (`nama_2019` berbeda dengan `Nama_2019`).
     - Menggunakan gaya *snake_case* dan tidak boleh menggunakan *reserved keywords* Python (seperti `if`, `while`, `for`).

2. **Tipe Data Skalar (Dasar)**.
   - **`str` (String)**: Menyimpan teks data pelanggan, seperti `nama_2019 = "Anita"` dan `kode_promo_2019 = "HEMAT10"`.
   - **`int` (Integer)**: Menyimpan bilangan bulat untuk jumlah barang (`jumlah_barang_2019 = 4`) dan representasi bitmask bitwise (`BIT_MEMBER_2019 = 1`).
   - **`float` (Float)**: Menyimpan bilangan desimal untuk nilai total belanja (`total_belanja_2019 = 250000.0`), persentase diskon, dan rata-rata harga.
   - **`bool` (Boolean)**: Menyimpan nilai kebenaran `True` atau `False` untuk evaluasi kondisi (contoh: `is_member_2019 = True`, `syarat_belanja_2019 = True`).

3. **Tipe Data Koleksi (Struktur Data)**.
   - **`dict` (Dictionary)**: Pasangan *key-value* untuk menyimpan daftar katalog produk `produk_2019` (contoh: `"1": {"nama": "Sabun", "harga": 5000}`).
   - **`list` (List)**: Menyimpan kumpulan data berurutan untuk daftar kode promo resmi `daftar_promo_2019 = ["HEMAT10", "HEMAT20", "GRATISONGKIR"]` dan riwayat daftar belanjaan.

4. **Konversi Tipe Data (*Type Casting*)**:
   - **Implicit Casting**: Otomatis dilakukan oleh Python, seperti hasil pembagian `total_belanja_2019 / jumlah_barang_2019` yang otomatis menjadi `float`.
   - **Explicit Casting**: Konversi manual menggunakan fungsi bawaan:
     - `int(input(...))` untuk mengonversi input jumlah barang menjadi integer.
     - `float(input(...))` untuk mengonversi input total belanja menjadi desimal.
     - `bool(...)` untuk mengonversi hasil operasi bitwise menjadi nilai boolean pada status hak akses.

## 2.2 Kode Program

### 2.2.1 Program "Boolean_2611532019"

```python
# Buat file dengan nama Boolean_2611532019
# Nama variabel ditambah 4 digit nim terakhir contoh: nilai_1234
# Deklarasi variabel dengan tipe data Boolean
is_lulus_2019= True

# Menggunakan Boolean
nilai_Fabhian_2019 = 70
nilai_Harland_2019 = 80
batas_lulus_2019 = 75

# Menantukan nilai Boolean dari kondisi
status_kelulusan_Fabhian_2019 = nilai_Fabhian_2019 >= batas_lulus_2019 # Hasilnya akan True
status_kelulusan_Harland_2019 = nilai_Harland_2019 >= batas_lulus_2019

print("=== Check Kelulusan ===")
print("Nilai: ",nilai_Fabhian_2019)
print("Apakah lulus?: ",status_kelulusan_Fabhian_2019)
if status_kelulusan_Fabhian_2019 == True :
    print("Selamat, Fabhian anda lulus!")
else:
    print("Maaf, Fabhian anda tidak lulus")

print("Nilai: ",nilai_Harland_2019)
print("Apakah lulus?: ",status_kelulusan_Harland_2019)
if status_kelulusan_Harland_2019 == True :
    print("Selamat, Harland Anda lulus!")
else:
    print("Maaf, Harland Anda tidak lulus")
```

**Penjelasan :**
1. **Tipe Data Boolean**: Menyimpan nilai kebenaran (True atau False) untuk menentukan status kelulusan berdasarkan nilai yang dimasukkan. 
2. **Operator Perbandingan:**: Membandingkan variabel nilai dengan batas kelulusan untuk menghasilkan kondisi True (jika memenuhi syarat) atau False (jika tidak memenuhi syarat).
3. **Pengondisian (if-else):**: Menyeleksi logika kelulusan. Jika status bernilai True, program mencetak pesan kelulusan ("Selamat, [nama] Anda lulus!"), dan jika False, program mencetak pesan tidak lulus ("Maaf, [nama] Anda tidak lulus").

### 2.2.2 Program "Input_2611532019"

```python
panjang_2019 = int(input("Masukkan nilai panjang: "))
lebar_2019 = int(input("Masukkan nilai lebar: "))
luas_2019 = (panjang_2019) * (lebar_2019)
print("Luas = ",luas_2019)
```

**Penjelasan :**

2. **Konversi Tipe Data int()**: Mengubah masukan yang awalnya bertipe string menjadi tipe data integer agar nilai panjang dan lebar dapat dihitung secara matematis.
3. **Perhitungan dan Output**: Menghitung luas persegi panjang dengan mengalikan nilai panjang_2019 dan lebar_2019, lalu menampilkan hasilnya ke layar menggunakan fungsi print().Contohnya panjang = 10 dan lebar = 20 hasil nya 200

### 2.2.3 Program "Konstanta_2611532019"

```python
phi = 3.14
jarijari_2019 = float(input("Masukkan jari-jari lingkaran : "))
keliling_2019 = float(2*phi*jarijari_2019)
luas_2019 = phi*jarijari_2019*jarijari_2019
print(f"jadi keliling dari lingkaran nya adalah : {keliling_2019}")
print(f"jadi luas dari lingkaran nya adalah : {luas_2019}")
```

**Penjelasan :**

### 2.2.4 Program "Numeric_2611532019"

```python
#create a variable with integer value
a_2019 = 100
print("The type of variable having value",a_2019,"is",type(a_2019))

#create variable with float value
b_2019 = 10.2345
print("The type of variable having value",b_2019,"is",type(b_2019))

#create variable with complex value
c_2019 = 100 + 3j
print("The type of variable having value",c_2019,"is",type(c_2019))
```

**Penjelasan :**
1. **Tipe Data Numeric**: Membuat variabel dengan berbagai tipe data angka, yaitu integer (int) untuk bilangan bulat, float (float) untuk bilangan desimal, dan complex (complex) untuk bilangan kompleks/imajiner.
2. **Fungsi type()**: Digunakan untuk mengecek dan memastikan jenis tipe data dari suatu variabel secara otomatis.
3. **Mencetak Output (print())**: Menampilkan nilai dari variabel beserta tipe datanya ke layar terminal.

```python
domisili_2019 = """
Kampung Dalam, Pauh,
Pasar Baru, Padang,
Sumatera Barat,
Indonesia """

tokenpraktikum_2019 = (20 + 19j)

print("===== SISTEM REGISTRASI PRAKTIKUM ALPRO 2026 =====")
nama_2019 = str(input("Masukkan Nama anda : "))

while True:
    jeniskelamin_2019 = str(input("Masukkan jenis kelamin anda (L/P) : ")).lower()
    if jeniskelamin_2019 in ["l","p"]:
        break
    else :
        print("Hanya masukkan L/P")

umur_2019 = int(input("Masukkan umur anda : "))
skorawal_2019 = float(input("Masukkan skor awal anda : "))

print("\n===== DATA PRAKTIKUM DAN HASIL PEMERIKSAAN =====")
print(f"Nama : {nama_2019} | type class : {type(nama_2019).__name__}")
print(f"Domisili : {domisili_2019} | type class : {type(domisili_2019).__name__}")
print(f"Umur : {umur_2019} | type class : {type(umur_2019).__name__}")
print(f"Skor tes awal : {skorawal_2019} | type class : {type(umur_2019).__name__}")
print(f"ID Token Sinyal : {tokenpraktikum_2019} | type class : {type(tokenpraktikum_2019).__name__}")

print("===== STATUS KELULUSAN PRAKTIKUM =====")
print("Batas minimum nilai : 75") 
if skorawal_2019 >= 75 :
    skorawal_2019 = True
    print(f"Status Kelulusan : {skorawal_2019} | type class : {type(skorawal_2019).__name__}")
else :
    skorawal_2019 = False
    print(f"Status Kelulusan : {skorawal_2019} | type class : {type(skorawal_2019).__name__}")
```


### **A. Penjelasan Tipe Data pada Program**

1. **`str` (String)**:
   - **Variabel**: `domisili_2019`, `nama_2019`, `jeniskelamin_2019`.
   - **Fungsi**: Menyimpan data berupa teks. Pada variabel `domisili_2019`, digunakan *multiline string* (`"""..."""`) untuk menyimpan alamat lengkap yang terdiri dari beberapa baris teks.

2. **`complex` (Bilangan Kompleks)**:
   - **Variabel**: `tokenpraktikum_2019 = (20 + 19j)`.
   - **Fungsi**: Menyimpan nilai bilangan kompleks yang memiliki bagian riil (`20`) dan bagian imajiner (`19j`).

3. **`int` (Integer)**:
   - **Variabel**: `umur_2019`.
   - **Fungsi**: Menyimpan data angka bulat yang diperoleh dari konversi manual input pengguna `int(input(...))`.

4. **`float` (Floating-Point)**:
   - **Variabel**: `skorawal_2019`.
   - **Fungsi**: Menyimpan nilai angka desimal/pecahan dari skor tes awal pengguna melalui `float(input(...))`.

5. **`bool` (Boolean)**:
   - **Variabel**: `skorawal_2019` (pada blok percabangan `if-else`).
   - **Fungsi**: Menyimpan nilai kebenaran (`True` atau `False`). Variabel `skorawal_2019` mengalami *re-assignment* (perubahan tipe data) dari `float` menjadi `bool` berdasarkan evaluasi kondisi nilai $\ge 75$.

---

### **B. Penjelasan Alur Program (Step-by-Step)**

1. **Inisialisasi Variabel Awal**:
   - Program mendefinisikan variabel `domisili_2019` menggunakan *multiline string* dan `tokenpraktikum_2019` dengan tipe data *complex*.

2. **Input Data Mahasiswa**:
   - Meminta input `nama_2019` (di-cast ke `str`).
   - **Perulangan Validasi (`while True`)**: Meminta input `jeniskelamin_2019`. Jika input berupa `"l"` atau `"p"` (diubah ke huruf kecil dengan `.lower()`), program mengeksekusi `break` untuk keluar dari loop. Jika salah, program menampilkan pesan `"Hanya masukkan L/P"` dan mengulang input.
   - Meminta input `umur_2019` (dikonversi ke `int`) dan `skorawal_2019` (dikonversi ke `float`).

3. **Output Data dan Pemeriksaan Tipe Class**:
   - Program mencetak seluruh data yang telah diinput beserta nama tipe datanya menggunakan fungsi `type(...).__name__`.

4. **Evaluasi Status Kelulusan (`if-else`)**:
   - Program mengecek kondisi apakah `skorawal_2019 >= 75`.
   - Jika **`True`** (nilai $\ge 75$), variabel `skorawal_2019` diubah nilainya menjadi boolean `True`, lalu dicetak status kelulusan beserta tipe class terbarunya (`bool`).
   - Jika **`False`** (nilai $< 75$), variabel `skorawal_2019` diubah nilainya menjadi boolean `False`, lalu dicetak status kelulusannya.
