### 2.2.1 Program "Boolean_2611532019"

```python
# Buat file dengan nama Boolean_2611532019
# Nama variabel ditambah 4 digit nim terakhir contoh: nilai_1234
# Deklarasi variabel dengan tipe data Boolean
is_lulus_2019= True

# Menggunakan Boolean
nilai_Fabhian_2019 = 70
nilai_Harland_2019 = 80
batas_lulus_2019 = 75

# Menantukan nilai Boolean dari kondisi
status_kelulusan_Fabhian_2019 = nilai_Fabhian_2019 >= batas_lulus_2019 # Hasilnya akan True
status_kelulusan_Harland_2019 = nilai_Harland_2019 >= batas_lulus_2019

print("=== Check Kelulusan ===")
print("Nilai: ",nilai_Fabhian_2019)
print("Apakah lulus?: ",status_kelulusan_Fabhian_2019)
if status_kelulusan_Fabhian_2019 == True :
    print("Selamat, Fabhian anda lulus!")
else:
    print("Maaf, Fabhian anda tidak lulus")

print("Nilai: ",nilai_Harland_2019)
print("Apakah lulus?: ",status_kelulusan_Harland_2019)
if status_kelulusan_Harland_2019 == True :
    print("Selamat, Harland Anda lulus!")
else:
    print("Maaf, Harland Anda tidak lulus")
```

**Penjelasan :**
1. **Tipe Data Boolean**: Menyimpan nilai kebenaran (True atau False) untuk menentukan status kelulusan berdasarkan nilai yang dimasukkan. 
2. **Operator Perbandingan:**: Membandingkan variabel nilai dengan batas kelulusan untuk menghasilkan kondisi True (jika memenuhi syarat) atau False (jika tidak memenuhi syarat).
3. **Pengondisian (if-else):**: Menyeleksi logika kelulusan. Jika status bernilai True, program mencetak pesan kelulusan ("Selamat, [nama] Anda lulus!"), dan jika False, program mencetak pesan tidak lulus ("Maaf, [nama] Anda tidak lulus").

### 2.2.2 Program "Input_2611532019"

```python
panjang_2019 = int(input("Masukkan nilai panjang: "))
lebar_2019 = int(input("Masukkan nilai lebar: "))
luas_2019 = (panjang_2019) * (lebar_2019)
print("Luas = ",luas_2019)
```

**Penjelasan :**
1. **Fungsi input()**: Digunakan untuk menerima masukan (input) teks dari pengguna berupa nilai panjang dan lebar persegi panjang.
2. **Konversi Tipe Data int()**: Mengubah masukan yang awalnya bertipe string menjadi tipe data integer agar nilai panjang dan lebar dapat dihitung secara matematis.
3. **Perhitungan dan Output**: Menghitung luas persegi panjang dengan mengalikan nilai panjang_2019 dan lebar_2019, lalu menampilkan hasilnya ke layar menggunakan fungsi print().Contohnya panjang = 10 dan lebar = 20 hasil nya 200

### 2.2.3 Program "Konstanta_2611532019"

```python
phi = 3.14
jarijari_2019 = float(input("Masukkan jari-jari lingkaran : "))
keliling_2019 = float(2*phi*jarijari_2019)
luas_2019 = phi*jarijari_2019*jarijari_2019
print(f"jadi keliling dari lingkaran nya adalah : {keliling_2019}")
print(f"jadi luas dari lingkaran nya adalah : {luas_2019}")
```

**Penjelasan :**
1. **Fungsi input()**: Digunakan untuk menerima masukan (input) teks dari pengguna berupa nilai panjang dan lebar persegi panjang.
2. **Konversi Tipe Data int()**: Mengubah masukan yang awalnya bertipe string menjadi tipe data integer agar nilai panjang dan lebar dapat dihitung secara matematis.
3. **Perhitungan dan Output**: Menghitung luas persegi panjang dengan mengalikan nilai panjang_2019 dan lebar_2019, lalu menampilkan hasilnya ke layar menggunakan fungsi print().Contohnya panjang = 10 dan lebar = 20 hasil nya 200

### 2.2.4 Program "Numeric_2611532019"

```python
#create a variable with integer value
a_2019 = 100
print("The type of variable having value",a_2019,"is",type(a_2019))

#create variable with float value
b_2019 = 10.2345
print("The type of variable having value",b_2019,"is",type(b_2019))

#create variable with complex value
c_2019 = 100 + 3j
print("The type of variable having value",c_2019,"is",type(c_2019))
```

**Penjelasan :**
1. **Tipe Data Numeric**: Membuat variabel dengan berbagai tipe data angka, yaitu integer (int) untuk bilangan bulat, float (float) untuk bilangan desimal, dan complex (complex) untuk bilangan kompleks/imajiner.
2. **Fungsi type()**: Digunakan untuk mengecek dan memastikan jenis tipe data dari suatu variabel secara otomatis.
3. **Mencetak Output (print())**: Menampilkan nilai dari variabel beserta tipe datanya ke layar terminal.

### 2.2.5 Program "String_2611532019"

```python
nama_2019 = "Fabhian"
jenis_kelamin_2019 = 'L'
alamat_2019 = """
    Dekat Parewa Coffee Unand Padang, Koto Luar,
    Pasar Baru, Padang,
    Sumatera Barat, Indonesia
"""

agama_2019 = "Islam"
print("Nama saya adalah: ", nama_2019)
print("Jenis kelamin saya adalah: ", jenis_kelamin_2019)
print("Alamat saya adalah: ", alamat_2019)
print("Agama saya adalah: ", agama_2019)
```

**Penjelasan :**

2. **String Multi-Baris ("""...""")**: Digunakan pada variabel alamat_2019 untuk menyimpan teks alur alamat yang panjang dan terdiri dari beberapa baris sekaligus.
