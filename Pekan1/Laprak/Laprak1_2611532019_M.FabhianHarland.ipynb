### 2.2.1 Program "NamaKelas_NIM"
Pada Bagian ini, program digunakan untuk menampilkan sintaks dasar Python seperti komentar multi-baris ("""....""") dan perintah 'print()', termasuk mencetak baris kosong serta penggunaan *escape character* backslash ('\')

```python
# Program pertama 
"""
Komentar 
dua baris atau lebih 
pertemuan 1 pratikum alpro
"""

print("Hello World")
print()
print("program ini menghasilkan")
print("Empat baris output")
print("/ \\ // \\\\ /// \\\\")
print("program ini" + "dibuat oleh a\n" + "M.Fabhian Harland dan 2611532019")
```

**Penjelasan :**
1. **Komentar Multi-Baris ('"""..."""')**: Digunakan di bagian paling atas program untuk mencantumkan dokumentasi, komentar maupun identitas pembuat program tanpa diikutsertakan dalam eksekusi kode. 
2. **Fungsi 'Print()' dan baris Kosong**: Perintah 'print()' digunakan untuk mencetak teks atau output ke layar
3. **Karakter Escape ('\')**: Menggunakan *backslash* ('\') sebagai *escape character, '\n' untuk membuat baris baru (*new line*)


### 2.2.2 Membuat Repository GitHub
Setelah program dibuat dan diuji pada VS Code, proyek dapat dihubungkan dengan GitHub. Repository digunakan untuk menyimpan kode program secara daring.

Tahapan yang dilakukan yaitu:

1. Membuat repository baru pada GitHub.
2. Membuka folder proyek menggunakan VS Code.
3. Membuka terminal pada VS Code.
4. Menginisialisasi Git pada folder proyek.
5. Menambahkan file ke dalam staging.
6. Membuat commit.
7. Menghubungkan repository lokal dengan repository GitHub.
8. Melakukan push untuk mengunggah kode ke GitHub.

Contoh perintah yang digunakan:

```python
git init
git add .
git commit -m "Initial commit"
git branch -M main
git remote add origin https://github.com/username/nama-repository.git
git push -u origin main
```

Penjelasan:

git init digunakan untuk membuat repository Git pada folder proyek.
git add . digunakan untuk menambahkan seluruh perubahan ke staging area.
git commit digunakan untuk menyimpan perubahan dengan sebuah pesan.
git branch -M main digunakan untuk menentukan branch utama dengan nama main.
git remote add origin digunakan untuk menghubungkan repository lokal dengan repository GitHub.
git push digunakan untuk mengunggah kode dari komputer ke GitHub.
